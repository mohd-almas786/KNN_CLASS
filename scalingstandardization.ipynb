```python
# Importing necessary libraries
from sklearn.preprocessing import MinMaxScaler, StandardScaler, MaxAbsScaler
import numpy as np

# Sample dataset
data = np.array([[1, -1, 2],
                 [2, 0, 0],
                 [0, 1, -1]])

# 1. MinMaxScaler
# Scales features to a range, typically [0, 1]
min_max_scaler = MinMaxScaler()
data_minmax = min_max_scaler.fit_transform(data)

print("Original Data:")
print(data)
print("\nMinMax Scaled Data (Range [0, 1]):")
print(data_minmax)
```

```python
# 2. StandardScaler
# Scales features to have zero mean and unit variance
standard_scaler = StandardScaler()
data_standard = standard_scaler.fit_transform(data)

print("\nStandard Scaled Data (Zero mean, unit variance):")
print(data_standard)
```

```python

# 3. MaxAbsScaler
# Scales features to the range [-1, 1] by dividing by the maximum absolute value
max_abs_scaler = MaxAbsScaler()
data_maxabs = max_abs_scaler.fit_transform(data)

print("\nMaxAbs Scaled Data (Range [-1, 1]):")
print(data_maxabs)
```
